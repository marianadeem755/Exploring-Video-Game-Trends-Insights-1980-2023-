# <p style="width: 800px; border-radius: 20px; border: 5px solid #8B0038; padding: 25px; background-color: #f5ac0f; text-align: center; color: #8B0038; font-weight: 900; font-size: 34px; text-shadow: 1px 1px 3px rgba(139, 0, 56, 0.3); box-shadow: 0px 8px 16px rgba(0, 0, 0, 0.4);">Lets Impute the Missing Values of Rating Column using linear regression</p>  

```python
# Check for missing values in the 'Rating' column
if df['Rating'].isnull().sum() == 0:
    print("No missing values in the 'Rating' column. Skipping imputation.")
else:
    # Split the dataset into missing and non-missing values
    df_missing = df[df['Rating'].isnull()]
    df_non_missing = df[~df['Rating'].isnull()]

    # Identify relevant features and target variable
    features = ['Number of Reviews', 'Plays', 'Playing']
    target = 'Rating'

    # Prepare the data
    X = df_non_missing[features]
    y = df_non_missing[target]

    # Train the regression model
    regression_model = LinearRegression()
    regression_model.fit(X, y)

    # Predict missing values
    X_missing = df_missing[features]
    predicted_ratings = regression_model.predict(X_missing)

    # Round the predicted values of rating column to the nearest integer
    predicted_ratings_rounded = [round(rating) for rating in predicted_ratings]

    # Impute missing values
    df_missing['Rating'] = predicted_ratings_rounded

    # Combine the imputed missing values with non-missing values
    df_imputed = pd.concat([df_non_missing, df_missing])

    # Assign it back to the original dataframe
    df['Rating'] = df_imputed['Rating']
```

>* In this Dataset the Rating column contains the missing values which is of about 85% so, I used the Linear Regression to impute the missing values of the Rating Column
>* As the Missing percentage of Rating is greater than 50-80% percent but I don't prefer to drop the Column because The rating column has great importance in vedio games dataset as It is necessary to take insights about how much rating the each vedio game have

```python
# Plot release year vs. number of reviews
df = df[df['Release Date'] != 'releases on TBD']
df['Release Date'] = pd.to_datetime(df['Release Date'])
df['Release Year'] = df['Release Date'].dt.year

reviews_by_year = df.groupby('Release Year')['Number of Reviews'].sum().reset_index()
valid_years = df['Release Year'].unique()
reviews_by_year = reviews_by_year[reviews_by_year['Release Year'].isin(valid_years)]
valid_years = valid_years[(valid_years >= 1980) & (valid_years <= 2023)]

plt.figure(figsize=(18, 9))
plt.plot(reviews_by_year['Release Year'], reviews_by_year['Number of Reviews'],
         marker='o', linestyle='-', color='#8B0038', markerfacecolor='#f5ac0f',
         markeredgecolor='#8B0038')

for x, y in zip(reviews_by_year['Release Year'], reviews_by_year['Number of Reviews']):
    label = f'{y/1e6:.1f}M' if y >= 1e6 else f'{y/1e3:.0f}K'
    plt.text(x, y, label, ha='center', va='bottom', fontsize=9,
             fontweight='bold', color='#8B0038')

plt.title('Release Date vs. Number of Reviews', fontsize=18, fontweight='bold', color='#8B0038')
plt.xlabel('Release Year', fontsize=15, color='#8B0038')
plt.ylabel('Number of Reviews', fontsize=15, color='#8B0038')
plt.grid(True, alpha=0.25)
plt.xticks(valid_years, rotation=45, fontsize=10, color='#8B0038')
plt.yticks(fontsize=12, color='#8B0038')
plt.tight_layout()
plt.show()
```

```python
# Create a network plot for the top 20 video game titles by number of reviews
N = 20
top_titles = df.nlargest(N, 'Number of Reviews')
G = nx.DiGraph()

for title, reviews in zip(top_titles['Title'], top_titles['Number of Reviews']):
    G.add_node(title, size=reviews)

node_sizes = [G.nodes[n]['size'] * 30 for n in G.nodes]
pos = nx.spring_layout(G, seed=42)

plt.figure(figsize=(14, 10))
nx.draw_networkx_nodes(G, pos, node_size=node_sizes, node_color='#f5ac0f', alpha=0.85)
nx.draw_networkx_edges(G, pos, edge_color='#8B0038', alpha=0.2)

node_labels = {title: f"{title}\n({reviews})" for title, reviews in zip(top_titles['Title'], top_titles['Number of Reviews'])}
nx.draw_networkx_labels(G, pos, labels=node_labels, font_size=10, font_family='sans-serif', font_color='#8B0038')

plt.title(f'Top {N} Video Game Titles Network by Number of Reviews', fontsize=18, fontweight='bold', color='#8B0038')
plt.axis('off')
plt.tight_layout()
plt.show()
```
